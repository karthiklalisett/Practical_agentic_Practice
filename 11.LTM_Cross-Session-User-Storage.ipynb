#### 3. Long-Term Memory: Cross-Session User Storage (BaseStore)
The Core Problem It Solves
Short-term memory (checkpointers, sliding windows, and summarization nodes) only exists within a single thread_id.

When a user closes their chat session today and starts a brand-new session (thread_id="session_2") tomorrow, checkpointers start with a completely fresh state. Long-Term Memory allows the agent to recall user preferences, facts, and profile information across different threads.

How LangGraph's BaseStore Works
LangGraph provides an in-memory/persistent store abstraction (InMemoryStore, AsyncPostgresStore, etc.) that uses hierarchical namespaces and keys:

Namespace Tuple: e.g., ("users", "user_123", "preferences")

Key: e.g., "coding_style"

Value: A JSON dictionary storing facts or key-value attributes.
