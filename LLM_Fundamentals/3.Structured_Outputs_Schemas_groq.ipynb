```python
import os
from langchain_groq import ChatGroq
from pydantic import BaseModel, Field
from langchain_groq import ChatGroq
```

```python
# 1. Define Strict Pydantic Schema

class APIEndpointConfig(BaseModel):
    endpoint_name: str = Field(
        description="The URI path fro the endpoint (e.g. /api/v1/users)"
    )
    http_method: str = Field(
        description="HTTP method in uppercase: GET, POST, PUT, DELETE"
    )
    requires_auth: bool = Field(
        description="Whether JWT/API token authentication is required"
    )
    query_params: list[str] = Field(
        default=[], description="List of expected query parameter names"
    )
    response_status_code: int = Field(
        default=200, description="Default successful HTTP status code"
    )
```

```python
llm = ChatGroq(
    model="openai/gpt-oss-20b",
    temperature=0,
    api_key="Your_API_KEY_HERE"
)
```

```python
# 3. Bind the Pydantic schema using with_structured_output()
structured_llm = llm.with_structured_output(APIEndpointConfig)
```

```python
# --- Run Live Request ---
if __name__ == "__main__":
  user_prompt = "Create an API configuration for an authenticated analytics query endpoint that accepts start_date and end_date filtering."

  print(f"Sending prompt to Groq (Llama 3.3 70B)...")
  result = structured_llm.invoke(user_prompt)

  print("\n================ LIVE GROQ STRUCTURED OUTPUT ================")
  print(type(result))  # Will be an instance of APIEndpointConfig!
  print(f"Endpoint Path : {result.endpoint_name}")
  print(f"HTTP Method   : {result.http_method}")
  print(f"Auth Required : {result.requires_auth}")
  print(f"Query Params  : {result.query_params}")
  print(f"Status Code   : {result.response_status_code}")
```
